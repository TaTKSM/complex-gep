## $k=4$ (clipped)

```python
%%time
params = {'n': 100*1, 'd': 150*1, 'p': 200*1,
          'v_x': v_x, 'tau_x': 0j, 'v_w': 1.0, 'tau_w': 0.3 + 0.4j, 
          'f_tag': 'Monomial', 'f_params': {'k': 4}}

outlier_portion = 0.01
_test_arr = generate_complex_Gaussian_array(v=params['v_x'] * params['v_w'], tau=params['tau_x'] * params['tau_w'], 
                                            shape=(10**7,), rng=np.random.default_rng(0))
_cutoff = np.quantile(np.abs(_test_arr), q = 1 - outlier_portion, method='hazen')

runner = Runner(params=params, cutoff=_cutoff)
runner.run(n_mc=2000, n_jobs=n_jobs, seed=1234)
```

```python
idx = '10-a'
```

```python
%%time
runner.generate_plot(n_bins=n_bins, return_file_type='pdf', ext=idx, 
                     legend_anchor=(0.42, 0.08))
```
